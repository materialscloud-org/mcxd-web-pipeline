```python
"""
Script to add in the AiiDA db:
1) a final_structures group that matches with the frontend;
2) mc2d_id in the extras.
"""

import aiida
from aiida import orm
import aiida.common.exceptions
import json
from pathlib import Path

aiida.load_profile("mc2d")
```

```python
# Create the group (or load if it exists)
group_label = "final_structures"
try:
    group = orm.Group.collection.get(label=group_label)
except aiida.common.exceptions.NotExistent:
    group = orm.Group(label=group_label)
    group.store()
```

```python
MC2D_JSON_PATH = Path().resolve().parent.parent / "data" / "pbe-v1" / "base" / "base.json"

with MC2D_JSON_PATH.open("r") as fh:
    meta = json.load(fh)
```

```python
for mc2d_id, entry in meta.items():
    structure_uuid = entry['general']['structure_relaxed_uuid']
    node = orm.load_node(structure_uuid)
    group.add_nodes(node)
    node.base.extras.set("mc2d_id", mc2d_id)
```
